## Seq2Seq 란?

- 자연어 처리(NLP) 분야에서 널리 사용되는 모델 
- 입력 시퀀스를 받아 출력 시퀀스를 생성하는데 사용된다. 

- 두개의 `RNN(Recurrent Nerual Network)` 또는 `LSTM(Long Short-Term Memory)` 네트워크를 연결하여 입력 시퀀스를 처리하고 출력 시퀀스를 생성하는 것 





1. Encoder ( 인코더 )

- 인코더는 입력 시퀀스를 받아 고정된 길이의 컨텍스트 벡터(Context vector)로 인코딩 

2. Decoder ( 디코더 )

- 디코더는 인코더의 컨텍스트 벡터를 받아 출력 시퀀스를 생성

3. 컨텍스트 벡터 ( Context Vector )

- 컨텍스트 벡터는 입력 시퀀스의 정보를 요약하여 단일 벡터로 표현 

## 2. 인코더 ( Encoder )

- 입력 시퀀스를 받아 고정된 길이의 벡터로 인코딩 하는 역할
- 고정된 길이의 벡터를 컨텍스트 벡터(context vector)라고 하며, 디코더의 초기 은닉 상태로 사용된다. 
- 인코더는 일반적으로 RNN(Recurrent Neural Network), LSTM(Long Short-Term Memory), 또는 GRU(Gated Recurrent Unit) 네트워크로 구성
