## Seq2Seq

- Sequence to Sequence 
- NLP분야에서 복잡한 시퀀스 변환 작업을 해결하기 위해 개발 됨 
- 주로 기계 번역(machine translation)과 같은 자연어 변환 작업의 필요성에서 기인 

- 초기 기계 번역 모델들은 주로 통계적 기법에 의존, 통계적 기계 번역(Statistical Machine Learning, SMT )은 언어 간 번역 규칙을 학습하기 위해 대규모 평행 코퍼스(parallel corpora)를 사용 
- 규칙 기반 접근법의 한계: 언어의 복잡한 문법과 의미적 뉘앙스를 포착하기 어렵고, 많은 예외 사항을 처리하는데 한계
- 데이터 의존성: 대규모 평행 코퍼스를 필요로하며, 데이터가 부족한 언어 쌍에서는 성능이 저하 

### RNN의 등장과 한계

- 순환신경망(Recurrent Neural Networks, RNN)은 시퀀스 데이터의 시간적 의존성을 처리할 수 있는 능력으로 NLP분야에서 주목받기 시작
- RNN은 입력 시퀀스를 순차적으로 처리하면서 이전의 정보를 기억하여 다음 단계로 전달, 그러나 RNN은 다음의 한계가 존재

1. 기울기 소실 문제 (vanishing gradient problem): 긴 시퀀스를 처리할 때, 기울기 소실로 인해 초반의 정보가 후반으로 전달되지 않는 문제가 발생
2. 장기 의존성 학습의 어려움: 
    - 중요한 정보가 시퀀스의 앞부분에 위치할 경우, 이를 기억하고 활용하는데 어려움이 있음. 

### LSTM과 GRU의 도입

- LSTM(Long Short-Term Memory)와 GRU(Gated Recurrent Unit)은 RNN의 기울기 소실 문제를 완화하기 위해 개발된 변형 모델 
- 이들은 게이트 메커니즘을 도입하여 장기 의존성을 학습할 수 있도록 개선
- 하지만 이들은 복잡한 시퀀스 변환 작업을 충분히 해결할 수 없었음

### Seq2Seq 모델의 제안

- 복잡한 시퀀스 변환 문제를 해결하기 위해 인코더-디코더(encoder-decoder)구조를 도입 
- LSTM이나 GRU는 Seq2Seq 모델의 요소로 자주 사용되며, Seq2Seq모델은 LSTM, GRU를 활용하는 특정 구조의 모델 



### LSTM

- LSTM(Long short-term memory) 은 RNN(Recurrent Neural Network)의 한 유형으로 장기적인 의존성을 처리하는데 효과적인 인공 신경망 구조 
- LSTM은 기존의 RNN의 단점을 보완, 긴 시퀀스의 데이터를 처리할 댸 발생하는 기울기 소실 문제를 해결 

### 장점

- 기울기 소실 문제 완화 
    - LSTM은 셀 상태를 유지함으로써 기울기 소실 문제를 완화하여 더 깊은 네트워크 구조에서도 효과적으로 학습 
- 게이트 메커니즘
    - 입력 게이트(input gate), 망각 게이트(forget gate), 출력 게이트(Output gate)라는 세가지 게이트를 통해 정보의 흐름을 조절, 이를 통해 불필요한 정보는 잊고 중요한 정보는 유지하여 학습의 효율성을 높인다 
- 병렬 처리의 어려움을 해결
    - 전통적인 RNN은 순차적인 데이터 처리로 인해 병렬 처리에 한계
    - LSTM은 게이트 구조를 톷ㅇ해 병렬 처리를 보다 효율적으로 수행 

## GRU(Gated Recurrent Unit)

- LSTM과 마찬가지로 순환신경망(RNN)의 한 유형, LSTM보다 더 단순한 구조
- 연산 비용을 줄이면서도 비슷한 성능을 낼 수 있다. 

### 1. 업데이트 게이트(Update Gate)

- 현재 입력과 이전 은닉 상태를 기반으로 현재 시점의 은닉 상태를 업데이트할지를 결정 
- 이 게이트는 새로운 정보를 얼마나 유지할지를 제어 

### 2. 리셋 게이트(Reset Gate)

- 이전 은닉 상태를 얼마나 반영할지를 결정
- 리셋 게이트는 새로운 입력과 결합하여 새로운 정보를 생성하는 역할 



### 장점

- 단순한 구조 
- 적은 매개변수
    - 동일한 데이터셋에서 과적합(Overfitting)의 위험이 줄어듬 
- 효율적인 학습
    - 모델 훈련 시간이 줄어들고, 빠른 프로토 타이핑 및 실험을 가능하게 함

- 기억 장기 의존성 
    - 긴 시퀀스의 장기 의존성을 효과적으로 학습할 수 있어, 텍스트 처리, 시계열 예측 등의 다양한 작업에서 유용 
- 딥러닝 모델에서 이러한 장점들 덕분에 널리 사용되고 있음. 
- 특히, 실시간 시스템이나 모바일 환경과 같은 제한된 자원에서 효과적으로 사용 

## 4. Seq2Seq 장점

- 유연한 구조 
    - 기계번연(Machine Translation)
    - 텍스트 요약
    - 대화 생성 
    - 이미지 캡셔닝
